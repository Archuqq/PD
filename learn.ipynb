```python
import polars as pl
import pandas as pd
```

```python
df = pl.read_csv("123.csv")

def parse_date_col(name: str, alias: str) -> pl.Expr:
    c = pl.col(name).cast(pl.Utf8, strict=False).str.strip_chars()
    return (
        pl.coalesce([
            c.str.to_datetime("%Y-%m-%d %H:%M:%S", strict=False),
            c.str.to_datetime("%Y-%m-%d", strict=False),
            c.str.to_datetime("%m/%d/%Y", strict=False),   
            c.str.to_datetime("%d.%m.%Y", strict=False),
        ])
        .dt.date()
        .alias(alias)
    )

df = df.with_columns(parse_date_col("defect_date", "d_new"))

df.select("defect_date", "d_new").head()

df = df.with_columns([
    pl.col("d_new").dt.year().alias("year"),
    pl.col("d_new").dt.month().alias("month"),
    pl.col("d_new").dt.day().alias("day"),
])

df.select("defect_date", "d_new", "year", "month", "day")
```
